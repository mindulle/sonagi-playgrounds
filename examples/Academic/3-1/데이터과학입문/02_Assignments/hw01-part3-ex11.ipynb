```python
# a) List comprehension
# 'data science'의 각 문자를 대문자로 바꾼 리스트 (공백 제외)
text = "data science"
upper_chars = [char.upper() for char in text if char != " "]
print(upper_chars)

# b) Dict comprehension
# ['a', 'as', 'bat', 'car', 'dove', 'python'] 리스트를 이용해
# {단어: 단어길이} 딕셔너리 생성 (길이가 2 초과인 것만)
words = ['a', 'as', 'bat', 'car', 'dove', 'python']
word_dict = {w: len(w) for w in words if len(w) > 2}
print(word_dict)

# c) Set comprehension
# 위 words 리스트에서 단어의 길이를 담은 셋 생성
len_set = {len(w) for w in words}
print(len_set)
```
