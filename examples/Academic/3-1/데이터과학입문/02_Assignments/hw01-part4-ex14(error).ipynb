```python
# a) Element-wise operations
a = np.array([1, 2, 3])
b = np.array([10, 20, 30])
print(f"Add: {a + b}, Mul: {a * b}, Exp: {np.exp(a)}")

# b) Broadcasting
# 3x3 매트릭스에 스칼라 10 더하기
m33 = np.ones((3, 3))
print(f"Broadcast Scalar:\n{m33 + 10}")

# (3,) 배열을 (3,1)로 reshape하여 (3,3) 매트릭스의 각 행에 더하기
row_vec = np.array([1, 2, 3]).reshape(3, 1)
print(f"Broadcast Column Vector:\n{m33 + row_vec}")
```
