```python
# a) Lambda
celsius_to_fahrenheit = lambda c: c * 9/5 + 32
temps = [0, 20, 37, 100]
print(list(map(celsius_to_fahrenheit, temps)))

# b) safe_divide
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return float('inf')
    except TypeError:
        raise TypeError("Custom Message: Inputs must be numeric.")

# 테스트
print([safe_divide(10, 2), safe_divide(10, 0)])
# safe_divide(10, '2') 는 에러를 발생시키므로 별도로 확인하세요.

# c) Sort with lambda
students = [{"name":"Alice","grade":88}, {"name":"Bob","grade":95}, {"name":"Carol","grade":72}]
sort_grade = sorted(students, key=lambda x: x['grade'], reverse=True)
sort_name = sorted(students, key=lambda x: x['name'])
print(f"Grade Desc: {sort_grade}\nName Asc: {sort_name}")
```
