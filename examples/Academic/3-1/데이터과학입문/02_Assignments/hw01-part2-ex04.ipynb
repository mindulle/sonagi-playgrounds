```python
# --- 기본 루프 (for & if/elif/else) ---
for i in range(1, 51):
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)

# --- Bonus: Single list comprehension ---
# 리스트 컴프리헨션 내에서 조건문을 중첩하여 사용합니다.
fizzbuzz_bonus = ["FizzBuzz" if i%15==0 else "Fizz" if i%3==0 else "Buzz" if i%5==0 else i for i in range(1, 51)]
print(fizzbuzz_bonus)
```
