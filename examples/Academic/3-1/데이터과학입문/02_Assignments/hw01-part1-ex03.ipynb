```python
# a) Arithmetic [cite: 34]
# 예측: 17 % 5 = 2, 2 ** 10 = 1024, 17 // 5 = 3, -17 // 5 = -4
print(17 % 5, 2 ** 10, 17 // 5, -17 // 5)
# // 연산자는 내림(floor)을 수행하므로, -3.4를 내림하면 -4가 됩니다. [cite: 34]

# b) Truthiness [cite: 35]
# 예측: False, False, False, False, True, True
print(bool(0), bool(""), bool([]), bool(None), bool(" "), bool([0]))

# c) Short-circuit [cite: 36]
# (i) True이므로 뒤의 1/0은 실행되지 않고 바로 True 반환
# (ii) False이므로 뒤의 1/0은 실행되지 않고 바로 False 반환
# (iii) None은 Falsy하므로 뒤의 "default"가 반환됨
print(True or (1/0))
print(False and (1/0))
print(None or "default")
```
