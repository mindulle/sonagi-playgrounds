```python
# a) Collatz sequence
n = 27
steps = 0
while n != 1:
    if n % 2 == 0:
        n = n // 2
    else:
        n = 3 * n + 1
    steps += 1
print(f"Steps to reach 1: {steps}") # 27로 시작하면 111단계가 나옵니다.

# b) Sentinel loop
numbers = [4, 7, -2, 15, -8, 3, -1, 9, 0, 6]
total_sum = 0
idx = 0
while idx < len(numbers):
    val = numbers[idx]
    if val == 0: # 0을 만나면 중단 (Sentinel)
        break
    if val > 0: # 양수만 합산
        total_sum += val
    idx += 1
print(f"Final sum: {total_sum}")
```
