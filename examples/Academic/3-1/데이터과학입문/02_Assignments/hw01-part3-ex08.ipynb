```python
# a) Nested tuple & indexing
nested_tup = (4, 5, 6, (7, 8))
# 8을 추출하려면?
print(nested_tup[3][1]) #

# b) Unpacking with *rest
values = (10, 20, 30, 40, 50)
a, b, *rest = values
print(f"a: {a}, b: {b}, rest: {rest}") # rest는 리스트로 반환됩니다.

# c) Swap logic
x, y = 1, 2
x, y = y, x # 튜플 언패킹을 이용한 교환
print(f"x: {x}, y: {y}")
```
