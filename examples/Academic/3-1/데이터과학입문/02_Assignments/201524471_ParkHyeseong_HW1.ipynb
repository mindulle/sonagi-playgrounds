```python
## [Exercise 5] While Loops
# a) Collatz sequence
n, steps = 27, 0
while n != 1:
    n = n // 2 if n % 2 == 0 else 3 * n + 1
    steps += 1
print(f"Collatz steps: {steps}")

# b) Sentinel loop
numbers = [4, 7, -2, 15, -8, 3, -1, 9, 0, 6]
total, idx = 0, 0
while idx < len(numbers) and numbers[idx] != 0:
    if numbers[idx] > 0: total += numbers[idx]
    idx += 1
print(f"Sentinel sum: {total}")
```

```python
## [Exercise 6] Function Design
def describe_number(n):
    """Returns the positivity and parity of an integer."""
    if n == 0: return "zero"
    return f"{'positive' if n > 0 else 'negative'} {'even' if n % 2 == 0 else 'odd'}"

def word_stats(text, sep=" "):
    """Returns statistics of words in a string."""
    words = text.split(sep)
    return {
        'word_count': len(words),
        'unique_words': len(set(words)),
        'longest_word': max(words, key=len),
        'avg_length': round(sum(len(w) for w in words)/len(words), 2)
    }

def running_stats(*nums):
    """Returns basic stats for variable arguments."""
    if not nums: return None
    return (min(nums), max(nums), sum(nums), sum(nums)/len(nums))

# Testing
print(describe_number(-5))
print(word_stats("the quick brown fox jumps over the lazy dog"))
print(running_stats(1, 2, 3, 4, 5))
```

```python
## [Exercise 7] Lambda Functions & Exception Handling
# a) Lambda: Celsius to Fahrenheit
celsius_to_fahrenheit = lambda c: c * 9/5 + 32
temps = [0, 20, 37, 100]
# Use map() and convert to list for printing
print(f"Fahrenheit Temperatures: {list(map(celsius_to_fahrenheit, temps))}")

# b) safe_divide(a, b)
def safe_divide(a, b):
    """Returns a / b with error handling for zero division and invalid types."""
    try:
        return a / b
    except ZeroDivisionError:
        return float('inf')
    except TypeError:
        # Raising a custom message as per requirement
        raise TypeError("Custom Message: Both inputs must be numeric values.")

# Demonstrate all three cases
print(f"Case 1 (Normal): {safe_divide(10, 2)}")
print(f"Case 2 (Zero Division): {safe_divide(10, 0)}")

# Case 3 (Type Error): This will raise an exception, so we wrap it in a try-block to show the message
try:
    print(f"Case 3 (Invalid Type): {safe_divide(10, '2')}")
except TypeError as e:
    print(e)

# c) Sort with lambda
students_list = [{"name":"Alice","grade":88}, {"name":"Bob","grade":95}, {"name":"Carol","grade":72}]
# (i) Sort by grade descending
sort_grade = sorted(students_list, key=lambda x: x['grade'], reverse=True)
# (ii) Sort by name alphabetically
sort_name = sorted(students_list, key=lambda x: x['name'])

print(f"Sorted by Grade (Desc): {sort_grade}")
print(f"Sorted by Name (Asc): {sort_name}")
```

```python
## [Exercise 8] Tuples & Unpacking

# a) Unpack & mutate
point = (3.5, -2.1, 0.0)
x, y, z = point
print(f"Unpacked values -> x: {x}, y: {y}, z: {z}")

# Trying to mutate the tuple
try:
    point[0] = 99
except TypeError as e:
    print(f"Mutation Error: {e}")
# Explanation: Tuples are immutable, meaning their elements cannot be changed after creation. 04 - Data Structure I.pptx]

# b) Extended unpacking
# Format: name, year, major, scores (list)
record = ("Hong Gildong", 2021, "Computer Science", [89, 91, 78, 95])
name, year, major, scores = record

# Compute average score
avg_score = sum(scores) / len(scores)
print(f"Student: {name}, Major: {major}, Average Score: {avg_score:.2f}")

# c) Swap using tuple unpacking
a, b = 10, 20
print(f"Before swap: a = {a}, b = {b}")
a, b = b, a # One-line swap without a temporary variable 04 - Data Structure I.pptx]
print(f"After swap:  a = {a}, b = {b}")
```

```python
## [Exercise 9] List Operations & Comprehensions

# a) Mutation
temperatures = [22.1, 19.8, 25.3, 18.5, 30.0, 27.6]
temperatures.append(23.4)          # (i) Append
temperatures.insert(2, 21.0)       # (ii) Insert at index 2
temperatures.remove(18.5)          # (iii) Remove by value
temperatures.sort(reverse=True)    # (iv) Sort descending
print(f"Final Temperatures: {temperatures}")

# b) Slicing
numbers = list(range(1, 21))
print(f"First 5: {numbers[:5]}")           # (i) First 5
print(f"Last 5: {numbers[-5:]}")           # (ii) Last 5
print(f"Every third: {numbers[::3]}")      # (iii) Every third element
print(f"Reversed: {numbers[::-1]}")        # (iv) Reversed without in-place change

# c) Comprehensions
words = ["pandas", "numpy", "scikit", "matplotlib", "seaborn", "plotly"]
# (i) List of lengths
lengths = [len(w) for w in words]
# (ii) UPPER CASE if length > 6
long_words_upper = [w.upper() for w in words if len(w) > 6]
# (iii) Dict mapping word to length
word_len_dict = {w: len(w) for w in words}

print(f"Lengths: {lengths}")
print(f"Long words (>6): {long_words_upper}")
print(f"Word-Length Dict: {word_len_dict}")

# d) Nested List Comprehension: 5x5 Multiplication Table
# Creates a list of lists: [[1,2,3,4,5], [2,4,6,8,10], ...] 04 - Data Structure I.pptx]
multi_table = [[i * j for j in range(1, 6)] for i in range(1, 6)]

print("5x5 Multiplication Table:")
for row in multi_table:
    print(row) # Print each row on a separate line as required
```

```python
## [Exercise 10] Grade Book Operations

grade_book = {
    "Alice":  {"HW1": 88, "HW2": 92, "Midterm": 78},
    "Bob":    {"HW1": 75, "HW2": 80, "Midterm": 90},
    "Carol":  {"HW1": 95, "HW2": 89, "Midterm": 85},
    "David":  {"HW1": 60, "HW2": 70, "Midterm": 65},
}

# a) Safe access
# Retrieve Bob's Midterm using .get()
bob_midterm = grade_book["Bob"].get("Midterm", 0)
print(f"Bob's Midterm: {bob_midterm}")

# Handle KeyError gracefully for missing key "Eve"
try:
    eve_data = grade_book["Eve"]
except KeyError:
    print("Key 'Eve' not found in grade_book. Handling KeyError gracefully.")

# b) Iteration & Formatted Report
print("\n--- Student Average Report ---")
for name, scores in grade_book.items():
    # Calculate average of HW1, HW2, and Midterm
    avg_score = sum(scores.values()) / len(scores)
    print(f"Student: {name:<7} | Average Score: {avg_score:.1f}")

# c) Dict comprehension: Top students (Average >= 82)
top_students = {
    name: round(sum(scores.values()) / len(scores), 1) 
    for name, scores in grade_book.items() 
    if (sum(scores.values()) / len(scores)) >= 82
}
print(f"\nTop Students (Avg >= 82): {top_students}")

# d) Merging new data
new_student = {"Eve": {"HW1": 91, "HW2": 85, "Midterm": 93}}
grade_book.update(new_student) # or grade_book |= new_student
print(f"\nUpdated Grade Book (Added Eve): {list(grade_book.keys())}")
```

```python
## [Exercise 11] Set Operations

ds101 = {"Alice", "Bob", "Carol", "David", "Eve"}
ds201 = {"Bob", "Eve", "Frank", "Grace", "Alice"}

# a) Set operations (using operators, not loops)
# (i) Students in both courses (Intersection)
both = ds101 & ds201
# (ii) Students in DS101 only (Difference)
ds101_only = ds101 - ds201
# (iii) Students in either course (Union)
either = ds101 | ds201
# (iv) Students in exactly one course (Symmetric Difference)
one_course_only = ds101 ^ ds201

print(f"Students in both: {both}")
print(f"Students in DS101 only: {ds101_only}")
print(f"Students in either course: {either}")
print(f"Students in exactly one course: {one_course_only}")

# b) Set Comprehension
numbers = list(range(1, 31))
# Create a set of numbers divisible by 2 or 3
divisible_set = {n for n in numbers if n % 2 == 0 or n % 3 == 0}

print(f"\nSet of numbers divisible by 2 or 3:\n{divisible_set}")
print(f"Number of unique values: {len(divisible_set)}")
```

```python
## [Exercise 12] Writing & Reading a Text File

# a) Write: Create poem.txt using a context manager
poem_content = [
    "In data we trust, in code we write,\n",
    "With pandas and NumPy, we make things right.\n",
    "From rows and columns, insights arise,\n",
    "Data science opens up our eyes."
]

with open('poem.txt', 'w', encoding='utf-8') as f:
    f.writelines(poem_content)
print("poem.txt has been created successfully.")

# b) Read back and Analyze
with open('poem.txt', 'r', encoding='utf-8') as f:
    lines = f.readlines()
    
# (i) Total number of lines
total_lines = len(lines)

# (ii) Total word count
# Flatten all lines and split by whitespace to count words
all_words = []
for line in lines:
    all_words.extend(line.split())
total_words = len(all_words)

# (iii) The line with the most words
# Use max() with a key that counts words in each stripped line
longest_word_line = max(lines, key=lambda l: len(l.split())).strip()

print(f"Total Lines: {total_lines}")
print(f"Total Word Count: {total_words}")
print(f"Line with most words: \"{longest_word_line}\"")
```

```python
## [Exercise 13] CSV Pipeline Implementation
import csv

# Step 1 — Create: Use csv.writer to create students.csv
header = ["student_id", "name", "major", "gpa", "year"]
rows = [
    ["S001", "Lee Jiyeon",   "Computer Science",        3.8,  3],
    ["S002", "Park Minho",   "Statistics",              3.2,  1],
    ["S003", "Kim Soojin",   "Data Science",            3.9,  4],
    ["S004", "Choi Hyunwoo", "Mathematics",             2.8,  2],
    ["S005", "Jung Yuna",    "Computer Science",        3.5,  2],
    ["S006", "Yoon Taehyun", "Artificial Intelligence", 3.7,  3],
    ["S007", "Han Minji",    "Statistics",              2.9,  1],
    ["S008", "Oh Seungjae",  "Data Science",            4.0,  4]
]

with open('students.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    writer.writerow(header)
    writer.writerows(rows)
print("Step 1: students.csv created.")

# Step 2 — Read & Parse: Store as a list of dicts
students_data = []
with open('students.csv', 'r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        # Strip whitespace and convert GPA to float
        row = {k.strip(): v.strip() for k, v in row.items()}
        row['gpa'] = float(row['gpa'])
        row['year'] = int(row['year'])
        students_data.append(row)

# Step 3 — Analyze: Using only built-in Python
# (i) Average GPA (2 decimal places)
avg_gpa = sum(s['gpa'] for s in students_data) / len(students_data)
# (ii) Student with the highest GPA
top_student = max(students_data, key=lambda x: x['gpa'])
# (iii) Number of students per major (using a dict)
major_counts = {}
for s in students_data:
    major = s['major']
    major_counts[major] = major_counts.get(major, 0) + 1
# (iv) Students in Year 3 or 4
upper_years = [s['name'] for s in students_data if s['year'] in [3, 4]]
# (v) Students with GPA >= 3.5, sorted alphabetically
high_gpa_alpha = sorted([s for s in students_data if s['gpa'] >= 3.5], key=lambda x: x['name'])

print(f"\nAnalysis Results:")
print(f"  - Average GPA: {avg_gpa:.2f}")
print(f"  - Top Student: {top_student['name']} ({top_student['gpa']})")
print(f"  - Major Counts: {major_counts}")
print(f"  - Upper Years (3 or 4): {upper_years}")

# Step 4 — Write output: honor_roll.csv (GPA >= 3.5, sorted by GPA desc)
honor_roll = [s for s in students_data if s['gpa'] >= 3.5]
# Sort by GPA descending
honor_roll_sorted = sorted(honor_roll, key=lambda x: x['gpa'], reverse=True)

with open('honor_roll.csv', 'w', newline='', encoding='utf-8') as f:
    # Fieldnames: student_id, name, gpa
    writer = csv.DictWriter(f, fieldnames=["student_id", "name", "gpa"])
    writer.writeheader()
    for s in honor_roll_sorted:
        # Write only the specified columns
        writer.writerow({"student_id": s["student_id"], "name": s["name"], "gpa": s["gpa"]})
print("\nStep 4: honor_roll.csv created with filtered results.")
```
