```python
# a) list methods
colors = ['red', 'green', 'blue']
colors.append('yellow')
colors.insert(1, 'orange')
last = colors.pop() # 'yellow' 제거
colors.extend(['indigo', 'violet'])
print(colors)

# b) Sorting
numbers = [7, 2, 5, 1, 9]
numbers.sort() # 원본 수정
print(f"Sorted: {numbers}")

# c) Slicing
# [start:stop:step] 구조를 이해해야 합니다.
# 0부터 19까지의 리스트에서 [5, 8, 11, 14]를 추출하는 슬라이싱을 작성하세요.
nums = list(range(20))
print(nums[5:15:3]) # 5번 인덱스부터 15번 전까지 3칸씩
```
