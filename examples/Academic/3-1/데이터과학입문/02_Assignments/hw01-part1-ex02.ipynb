```python
# a) Assign and print [cite: 24]
my_int = 10
my_float = 3.14
my_bool = True
my_complex = 1 + 2j
my_none = None

for val in [my_int, my_float, my_bool, my_complex, my_none]:
    print(f"Value: {val}, Type: {type(val)}") # [cite: 25]

# b) Type conversion [cite: 26]
print(float(my_int))      # int -> float
print(int(3.9))           # float -> int (소수점 버림 확인)
print(int("123"))         # str -> int

try:
    int("python")         # ValueError 유도 [cite: 27]
except ValueError as e:
    print(f"Expected Error: {e}")

# c) String formatting [cite: 28]
name = "Data Science"
year = 2026
print("Introduction to %s - %d" % (name, year))     # (i) % formatting
print("Introduction to {} - {}".format(name, year)) # (ii) .format()
print(f"Introduction to {name} - {year}")           # (iii) f-string

# d) String methods [cite: 29]
sentence = "  python for data analysis  "
# 순차적으로 적용 [cite: 30]
s_strip = sentence.strip()
s_upper = s_strip.upper()
s_split = s_upper.split()
s_replace = s_upper.replace('PYTHON', 'PYTHON') # 이미 대문자이므로 결과 확인
print(f"Strip: '{s_strip}'\nUpper: '{s_upper}'\nSplit: {s_split}\nReplace: '{s_replace}'")
```
