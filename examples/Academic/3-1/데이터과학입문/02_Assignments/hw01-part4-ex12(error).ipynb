```python
import numpy as np

# a) Create from list
# [10, 20, 30, 40, 50] 리스트로 1D 배열 생성 및 타입 확인
arr_1d = np.array([10, 20, 30, 40, 50])
print(f"Shape: {arr_1d.shape}, Dtype: {arr_1d.dtype}") #

# b) Built-in creation functions
zeros_arr = np.zeros((3, 4))      # 3x4 영행렬
range_arr = np.arange(0, 21, 2)   # 0부터 20까지 2씩 증가
lin_arr = np.linspace(0, 1, 5)    # 0부터 1까지 균등하게 5분할
print(f"Zeros:\n{zeros_arr}\nRange: {range_arr}\nLinspace: {lin_arr}")

# c) Reshaping
# 0~11까지 12개 원소를 가진 1D 배열을 3x4 2D 배열로 변환
matrix = np.arange(12).reshape(3, 4)
print(f"Reshaped Matrix:\n{matrix}")
```
