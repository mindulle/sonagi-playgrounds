```python
import csv

# 데이터 생성
data = [
    ["name", "id", "gpa"],
    ["Alice", "20230001", "3.8"],
    ["Bob", "20230002", "3.2"],
    ["Charlie", "20230003", "4.1"],
    ["David", "20230004", "3.5"]
]

# 파일 쓰기
with open('students.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    writer.writerows(data)
print("students.csv 파일이 생성되었습니다.")
```
