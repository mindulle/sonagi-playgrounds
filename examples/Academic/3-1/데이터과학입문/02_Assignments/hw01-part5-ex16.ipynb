```python
# Step 2: Read & Parse
students_list = []
with open('students.csv', 'r', encoding='utf-8') as f:
    # DictReader를 쓰면 컬럼명을 키(key)로 하는 딕셔너리로 읽어와서 편리합니다.
    reader = csv.DictReader(f)
    for row in reader:
        # gpa는 계산을 위해 float으로 변환
        row['gpa'] = float(row['gpa'])
        students_list.append(row)

# Step 3: Analyze
gpas = [s['gpa'] for s in students_list]
avg_gpa = sum(gpas) / len(gpas)
# max 함수에 key 옵션을 주어 gpa가 가장 높은 딕셔너리를 찾습니다.
top_student = max(students_list, key=lambda x: x['gpa'])

print(f"Average GPA: {avg_gpa:.2f}")
print(f"Top Student: {top_student['name']} (ID: {top_student['id']})")
```
