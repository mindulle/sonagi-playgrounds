```python
# a) Dict creation & access
# 'apple': 5, 'banana': 3, 'cherry': 7 딕셔너리 생성
fruits = {'apple': 5, 'banana': 3, 'cherry': 7}
print(fruits.get('durian', 0)) # 키가 없을 때 기본값 반환

# b) Merge dicts (Python 3.9+)
d1 = {'a': 1, 'b': 2}
d2 = {'b': 3, 'c': 4}
d1.update(d2) # 또는 d3 = d1 | d2
print(d1) # 'b'의 값은 3으로 업데이트됨

# c) Sets
s1 = {1, 2, 3, 4, 5}
s2 = {4, 5, 6, 7, 8}
print(f"Union: {s1 | s2}")         # 합집합
print(f"Intersection: {s1 & s2}")  # 교집합
print(f"Difference: {s1 - s2}")    # 차집합
```
