```python
# a) describe_number
def describe_number(n):
    """숫자의 양수/음수 및 홀수/짝수 여부를 반환합니다.""" # Docstring
    if n == 0: return "zero"
    
    status = "positive" if n > 0 else "negative"
    parity = "even" if n % 2 == 0 else "odd"
    return f"{status} {parity}"

# 테스트 (5개 입력)
print([describe_number(x) for x in [10, -5, 0, 2, -8]])

# b) word_stats
def word_stats(text, sep=" "):
    """텍스트의 단어 통계를 딕셔너리로 반환합니다."""
    words = text.split(sep)
    # 리스트 컴프리헨션을 활용한 데이터 추출
    lengths = [len(w) for w in words]
    return {
        'word_count': len(words),
        'unique_words': len(set(words)),
        'longest_word': max(words, key=len),
        'avg_length': round(sum(lengths) / len(words), 2)
    }

print(word_stats("the quick brown fox jumps over the lazy dog"))

# c) running_stats
def running_stats(*numbers):
    """가변 인자를 받아 기초 통계량을 튜플로 반환합니다."""
    if not numbers: return None
    return (min(numbers), max(numbers), sum(numbers), sum(numbers)/len(numbers))

print(running_stats(10, 20, 30, 40))
```
